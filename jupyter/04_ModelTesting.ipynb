```python
import pandas as pd
import joblib
```

```python
model = joblib.load("../models/cloud_model.pkl")
feature_columns = joblib.load("../models/model_features.pkl")
```

```python
new_data = pd.DataFrame([{
    "application_type": "E-Commerce",
    "expected_users_per_day": 25000,
    "concurrent_users": 600,
    "storage_required_gb": 300,
    "deployment_region": "ap-south-1",
    "traffic_pattern": "High"
}])
```

```python
new_data = pd.get_dummies(new_data)
```

```python
new_data = new_data.reindex(columns=feature_columns, fill_value=0)
```

```python
prediction = model.predict(new_data)
```

```python
predicted_vcpu = prediction[0][0]
predicted_ram = prediction[0][1]

print("Predicted vCPU:", predicted_vcpu)
print("Predicted RAM:", predicted_ram, "GB")
```

```python
test_cases = pd.DataFrame([
    {
        "application_type": "Blog",
        "expected_users_per_day": 1000,
        "concurrent_users": 20,
        "storage_required_gb": 20,
        "deployment_region": "ap-south-1",
        "traffic_pattern": "Low"
    },
    {
        "application_type": "Healthcare",
        "expected_users_per_day": 40000,
        "concurrent_users": 700,
        "storage_required_gb": 500,
        "deployment_region": "us-east-1",
        "traffic_pattern": "High"
    }
])
```

```python
test_cases = pd.get_dummies(test_cases)
test_cases = test_cases.reindex(columns=feature_columns, fill_value=0)
predictions = model.predict(test_cases)
for i, pred in enumerate(predictions):
    print(f"Test Case {i+1}")
    print("Predicted vCPU:", pred[0])
    print("Predicted RAM:", pred[1], "GB")
    print("-" * 30)
```
